### Factorial


```python
def factorial(n: int) -> int:
    if n==0:
        return 1
    return n*factorial(n-1)
```

Here, each call to `factorial(n)` runs in $O(1)$ time, and there are $n+1$ total calls, so the runtime of the algorithm is $O(n)$.


```python
factorial(5)
```

The factorial function can also be implemented using a loop as follows. It is an incredibly inefficient algorithm, though.


```python
def factorial_using_loop(n: int) -> int:
    fact=1
    while n>0:
        fact*=n
        n-=1
    return fact

print(factorial_using_loop(5))
```

### English Ruler


```python
# English Ruler using Recursion
def draw_line(tick_length, tick_label=''):
    line = '-' * tick_length
    if tick_label:
        line += ' ' + tick_label
    print(line)

def draw_interval(center_length):
    if center_length > 0:
        draw_interval(center_length - 1)
        draw_line(center_length)
        draw_interval(center_length - 1)

def draw_ruler(num_inches, major_tick_length):
    draw_line(major_tick_length, '0')
    for inch in range(1, num_inches + 1):
        draw_interval(major_tick_length - 1)
        draw_line(major_tick_length, str(inch))
```

This algorithm runs in $O(2^c)$ where $c$ is the tick length. The number of lines printed by `draw_interval(c)` is one more than twice the number of lines printed by `draw_interval(c-1)`. It can be proved by induction that the total lines printed will be $2^c - 1$.


```python
draw_ruler(2, 5)
```

### Binary Search


```python
def binary_search(target: int, data: list, low: int, high: int) -> bool:
    if low > high:
        return False
    mid = (low + high) // 2
    if target == data[mid]:
        return True
    elif target > data[mid]:
        return binary_search(target, data, mid+1, high)
    else:
        return binary_search(target, data, low, mid-1)
```

All the statements in the `binary_search()` function run in $O(1)$ time, so the runtime of the algorithm is proportional to the number of calls to the function. For `data` of size $n$, this turns out to be $O(\log{n})$ because the array `data` will be repeatedly divided into half until $\dfrac{n}{2r} < 1$ where $r$ is the number of calls.


```python
data = [1, 4, 6, 2, 23, 75, 12, 3, 90]
target_1 = 4
target_2 = 100
low, high = 0, len(data)-1
print(binary_search(target_1, data, low, high))
print(binary_search(target_2, data, low, high))
```

However, this algorithm is a classic case of Tail Recursion. It uses $O(\log{n})$ memory because Python has no Tail Call Optimization. We can convert this (and any Tail Recursion) into an iterative algorithm using loops, which would reduce the addiional memory usage to $O(1)$.


```python
def binary_search_iterative(target: int, data: list, low: int, high: int) -> bool:
    while high>=low:
        mid = (low + high)//2
        if target == data[mid]:
            return True
        elif target > data[mid]:
            low = mid+1
        else:
            high = mid-1
    return False

binary_search_iterative(4, [1,3,5,6], 0, 3)
```

### Fibonacci
Returning the $n^{th}$ Fibonacci number


```python
def bad_fibonacci(n):
    if n<=1:
        return n
    return bad_fibonacci(n-1) + bad_fibonacci(n-2)
```

This is a terrible algorithm as it runs in $O(2^n)$ time. There are two recursive calls in each call to the `bad_fibonacci()` function, and each of them grows exponentially in the size they're operating on at each level. Let $c_n$ represent the number of calls for input $n$, then,
$c_0 = 1$

$c_1 = 1$

$c_2 = 1+c_0 + c_1 = 3$

$c_3 = 1 + c_1 + c_2 = 5$

$c_4 = 1 + c_2 + c_3 = 9$

$c_7 = 1 + c_5 + c_6 = 41$ and so on.

We can instead return a pair of Fibonacci numbers, so that a number already calculated at the previous level of recursion is passed up at the next level, avoiding the need for additional calculation.

The following algorithm runs in $O(n)$.


```python
def fibonacci(n):
    if n<=1:
        return (0, n)   # returns (0,1) on n=1, the first pair
    (a, b) = fibonacci(n-1)
    return (b, a+b)

fibonacci(5)    # returns 4rth and 5th Fibonacci numbers
```

### Recursive Sum of an Array
This computes the sum of the first $n$ numbers of the sequence $S$ (an array-like object) in $O(n)$ time.


```python
def recursive_sum(S, n):
    if n < 1:
        return 0
    return recursive_sum(S, n-1) + S[n-1]

recursive_sum([1,2,3,4,5,6,7], 7)
```

We can improve it further, by using **binary recursion**. We can split the list into half, and add each half recursively, as follows. This runs in the same $O(n)$ time, but uses only $O(\log n)$ additional memory, instead of $O(n)$ in the previous.


```python
# start and stop just allow use to sum a specific range, this doesn't affect the algorithm substantially, we could implement the same strategy in the previous sum algorithm similarly.

def binary_sum(S, start, stop):
    if stop<=start:
        return 0
    elif stop-1 == start:
        return S[start]
    else:
        mid = (start + stop)//2
        return binary_sum(S, start, mid) + binary_sum(S, mid, stop)

print(binary_sum([1,2,3,4,5,6,7], 0, 7))
```

### Reverse a Sequence
This reverses a Python list in place using recursion in $O(n)$ time.


```python
def reverse(S, start, stop):
    if start <= stop:
        S[start], S[stop-1] = S[stop-1], S[start]
        reverse(S, start+1, stop-1)

S=[1,2,3,4,5]
reverse(S, 0, 5)
print(S)
```

### Power
The following algorithms calculate $x^n$ for some integers $x$ and $n$. The first algorithm `power_bad()`, although not too bad, runs in $O(n)$ time.


```python
def power_bad(x: int, n: int) -> int:
    if n < 1:
        return 1
    return x * power_bad(x, n-1)
```

The above algorithm does that job, but has runtime and memory usage of $O(n)$ (because $O(n)$ records are simultaneously stored in memory due to a recursion depth of $O(n)$. This can be further improved with the following algorithm that runs in $O(\log{n})$ time and memory.


```python
def power(x: int, n: int, call_n=0) -> int:
    if n<1:
        return 1
    partial = power(x, n//2)    # Only compute half the power and then
    result = partial * partial  # square it
    if n%2 == 1:        # Additional factor for odd number power
        result *= x
    return result
```

```python
print(power(4, 4))
```

### Tower of Hanoi


```python
# Tower of Hanoi Problem

def move_disk(from_pole, to_pole):
    print(f"Move disk from {from_pole} to {to_pole}")

def hanoi(n, from_pole, to_pole, with_pole):
    if n == 1:
        move_disk(from_pole, to_pole)
    else:
        hanoi(n - 1, from_pole, with_pole, to_pole)
        move_disk(from_pole, to_pole)
        hanoi(n - 1, with_pole, to_pole, from_pole)
hanoi(3, 'A', 'C', 'B')
```

**Deep Flattening**
Given a nested list of integers, where each element is either an integer or another list (which can itself contain integers or lists), write a function to return a single "flat" list containing all the integers in their original order.

Example Input: `[1, [2, [3, 4], 5], 6, [7, 8]]`

Expected Output: `[1, 2, 3, 4, 5, 6, 7, 8]`


```python
# Functional Recursion
from typing import List, Any

def flatten_recursive(nested_list: List[Any]) -> List[int]:
    """
    Flattens a nested list using a standard recursive approach.
    Complexity: O(N) where N is the total number of integers.
    """
    flat_list = []

    for element in nested_list:
        if isinstance(element, list):
            # Recursive step: Extend the list with the flattened sub-list
            flat_list.extend(flatten_recursive(element))
        else:
            # Base case/Leaf node: Append the integer
            flat_list.append(element)

    return flat_list

# Usage
data = [1, [2, [3, 4], 5], 6, [7, 8]]
print(flatten_recursive(data))
```
