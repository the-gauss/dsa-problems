# 9. Palidrome Number
Refer to [Leetcode](https://leetcode.com/problems/palindrome-number/) for problem statement.

### Solution 1: The String Cast

**The Good**: It is highly Pythonic, concise, and correctly handles negative numbers because the minus sign is preserved in the string reversal ("-121" becomes "121-", which correctly evaluates to False).

**The Bad**: It is an entry-level shortcut. In an algorithmic interview, or a highly constrained environment, this approach is a failure. You are allocating extra memory to create two string objects. This takes $O(n)$ space, where $n$ is the number of digits. The core intent of this problem is to test your ability to manipulate integers mathematically, not your knowledge of Python string slicing.

```python
def is_palindrome_str(x: int) -> bool:
    return str(x)[::-1]==str(x)

tests = [121, -121, 10001, 10, 0, 200000000000000002]
for t in tests:
    print(t, is_palindrome_str(t))
```

### Solution 2: Using Integer Manipulation
This improves on the previous solution in that we don't need $O(n)$ space due to string usage. We are manipulating the integer itself.
Time Complexity: $O(\log_{10}(x))$ (proportional to the number of digits in $x$)
Space Complexity: $O(1)$

```python
def is_palindrome_int(x:int) -> bool:
    if x < 0:       # Negative numbers can't be palindromes
        return False
    original=x
    reverse=0
    while x/10 != 0:
        last = x%10
        x = x//10
        reverse = last + reverse * 10
    return reverse==original

for t in tests:
    print(t, is_palindrome_int(t))
```

### Solution 3: The Optimal Solution
This solution further improves using the fact that a palindrome number is essentially mirrored at the center. You don't need to perform calculations over the entire number and then compare it with the original number, you can simply reverse the second half of the number and compare it to the first.
Moreover, this dramatically reduces the risk of integer overflow in statically typed languages like Java or C++ (although it's not a major problem here for Python).

```python
def is_palindrome(x:int) -> bool:
    # Fast fail: All negative numbers are not palindromes.
    # Numbers ending in 0 (but not 0 itself) cannot be palindromes.
    if x<0 or (x % 10 == 0 and x != 0):
        return False

    reversed = 0

    # We only reverse until the reversed half is greater than or equal to the remaining half
    while x > reversed:
        reversed = (reversed * 10) + x % 10
        x //= 10

    # For even digit lengths, x == reverted.
    # For odd digit lengths, we drop the middle digit using reverted // 10.
    return x == reversed or x == reversed // 10

for t in tests:
    print(t, is_palindrome(t))
```
