```python
class Solution:
    def plusOne(self, digits: List[int]) -> List[int]:
        if self.is_all_nine(digits):
            for digit in digits:
                digit=0
            digits = digits.append(1)[::-1]
        elif digits[-1]==9:
            digits[-2]+=1
            digits[-1]=1
        else:
            digits[-1]+=1
        return


    
    def is_all_nine(self, digits: List[int]) -> List[int]:
        for digit in digits[:-1]:
            if digit != 9:
                return False
            return True

```
